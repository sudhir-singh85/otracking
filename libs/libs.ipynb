```python
from pyspark.sql.types import IntegerType, FloatType, DoubleType
from pyspark.sql.functions import to_timestamp,try_to_timestamp,substring, col
```

```python
def remove_Dups(df):
    print('Removing Duplicate values: ', end='')
    df_dup = df.dropDuplicates()
    print('Success!! ')
    return df_dup
```

```python
def handle_NULLs(df,columns):
    print('Replacing NULL values on String Columns with "Unknown" ' , end='')
    df_string = df.fillna('Unknown',subset= columns)
    print('Successs!! ')

    print('Replacing NULL values on Numeric Columns with "0" ' , end='')
    df_clean = df_string.fillna(0,subset = columns)
    print('Success!! ')

    return df_clean
```

```python
def convert_to_int(df,column):
    df_casted = df.withColumn(column,df[column].cast(IntegerType()))
    return df_casted
```

```python
def convert_to_float(df,column):
    df_casted = df.withColumn(column,df[column].cast(FloatType()))
    return df_casted
```

```python
def convert_to_double(df,column):
    df_casted = df.withColumn(column,df[column].cast(DoubleType()))
    return df_casted
```

```python
def convert_to_timestamp(df,column):
    df_casted = df.withColumn(column,to_timestamp(df[column]))
    return df_casted
 
```

```python
def convert_to_try_timestamp(df,column):
    df_casted = df.withColumn(column,try_to_timestamp(df[column]))
    return df_casted
```

```python
def convert_to_substring(df,column):
    df_casted = df.withColumn(column,substring(df[column], 1, 10))
    return df_casted
```

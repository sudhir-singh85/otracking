```python
base_path = f"{tracking_gd_path}/files"
```

```python
from pyspark.sql.functions import current_timestamp
df_dimbank = spark.read.csv(f"{base_path}/bankname.csv", header=True, inferSchema=True)
df_dimbank= df_dimbank.withColumn("ingestion_time", current_timestamp())
```

```python
df_dimbank.createOrReplaceTempView("temp_bank")
```

```python
%sql
merge into dev.tracking_gd.dimbank tgt
USING ( 
  SELECT DISTINCT * 
  FROM temp_bank 
) src 
on tgt.bankname = src.bankname
when not matched then
  insert (bankname, ingestion_time)
  VALUES(src.bankname,src.ingestion_time)
```

```python
def createTable_dimbank(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimbank
                        (
                            bankid long GENERATED ALWAYS AS IDENTITY,
                            bankname VARCHAR(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (bankname) LOCATION '{path}/dimbank';""")
```

```python
def createTable_dimpayment(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimpayment
                        (
                            paymentid long GENERATED ALWAYS AS IDENTITY,
                            payment_ref VARCHAR(100),
                            datekey int,
                            chequeno varchar(100),
                            details varchar(255),
                            ingestion_time TIMESTAMP
                    ) LOCATION '{path}/dimpayment';""")
```

```python
def createTable_dimtracking(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimtracking
                        (
                            trackingid long GENERATED ALWAYS AS IDENTITY,
                            shipmentno VARCHAR(100),
                            status varchar(100),
                            datekey int,
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (status) LOCATION '{path}/dimtracking';""")
```

```python
def createTable_dimproduct(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimproduct
                        (
                            productid long GENERATED ALWAYS AS IDENTITY,
                            category VARCHAR(100),
                            size numeric(18,2),
                            prouctvalue numeric(18,2),
                            weight numeric(18,2),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (category) LOCATION '{path}/dimproduct';""")
```

```python
def createTable_dimorganization(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimorganization
                        (
                            orgid long GENERATED ALWAYS AS IDENTITY,
                            name VARCHAR(100),
                            city VARCHAR(100),
                            state VARCHAR(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (state) LOCATION '{path}/dimorganization';""")
```

```python
def createTable_diminvoice(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.diminvoice
                        (
                            invoiceid long GENERATED ALWAYS AS IDENTITY,
                            invoiceno VARCHAR(100),
                            datekey int,
                            order_id long,
                            ingestion_time TIMESTAMP
                    ) LOCATION '{path}/diminvoice';""")
```

```python
def createTable_dimorder(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimorder
                        (
                            orderid long GENERATED ALWAYS AS IDENTITY,
                            orderno VARCHAR(100),
                            datekey int,
                            shipmentno VARCHAR(100),
                            vehicle_id long,
                            sender_id long,
                            receiver_id long,
                            ingestion_time TIMESTAMP
                    )LOCATION '{path}/dimorder';""")
```

```python
def createTable_dimvehicle(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimvehicle
                        (
                            vehicleid long GENERATED ALWAYS AS IDENTITY,
                            vehicleno VARCHAR(100),
                            carrriername VARCHAR(100),
                            carriercity VARCHAR(100),
                            carrierstate VARCHAR(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (carrierstate) LOCATION '{path}/dimvehicle';""")
```

```python
def createTable_factinvoice(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.factinvoice
                        (
                            factinvoiceid long GENERATED ALWAYS AS IDENTITY,
                            datekey int,
                            invoice_id long,
                            invoice_amt numeric(18,2),
                            tax_rate numeric(18,2),
                            tax_amount numeric(18,2),
                            paymentdays int,
                            customer_id long,
                            ingestion_time TIMESTAMP
                    )LOCATION '{path}/factinvoice';""")
```

```python
def createTable_factorder(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.factorder
                        (
                            factorderid long GENERATED ALWAYS AS IDENTITY,
                            datekey int,
                            order_id long,
                            product_id long,
                            noofitems int,
                            estimatedcost numeric(18,2),
                            deliverycost numeric(18,2),
                            ingestion_time TIMESTAMP
                    )LOCATION '{path}/factorder';""")
```

```python
def createTable_factpayment(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.factpayment
                        (
                            factpaymentid long GENERATED ALWAYS AS IDENTITY,
                            datekey int,
                            payment_id long,
                            paymenttype varchar(100),
                            payer_id long,
                            receiver_id long,
                            bank_id long,
                            paymentamount numeric(18,2),
                            ingestion_time TIMESTAMP
                    )LOCATION '{path}/factpayment';""")
```

```python
def createTable_dimdate(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_gd.dimdate
                        (
                            Datekey int,
                            Date DATE,
                            Year int,
                            Month int,
                            Day int,
                            DayName varchar(50),
                            MonthName varchar(50),
                            ingestion_time TIMESTAMP
                    )LOCATION '{path}/dimdate';""")
```

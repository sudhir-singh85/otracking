```python
def createTable_ordertracking(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS otracking.silver.ordertracking
                        (
                            trackingid int,
                            shipmentno VARCHAR(100),
                            orderstatus varchar(100),
                            trackingtime TIMESTAMP,
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (orderstatus) LOCATION '{path}/silver/ordertracking';""")
```

```python
def createTable_invoice(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS otracking.silver.invoice
                        (
                            invoiceno VARCHAR(100),
                            invoicedate VARCHAR(100),
                            orderno varchar(100),
                            invoiceamount varchar(100),
                            taxrate varchar(100),
                            taxamount varchar(100),
                            paymentdays varchar(100),
                            nameofcustomer varchar(255),
                            invoicecity varchar(100),
                            invoicestate varchar(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (invoicestate) LOCATION '{path}/silver/invoice';""")
```

```python
def createTable_order(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS otracking.silver.order
                        (
                            orderno VARCHAR(100),
                            orderdate timestamp,
                            orderweight numeric(18,2),
                            ordervalue numeric(18,2),
                            productcategory varchar(100),
                            productname varchar(100),
                            productsize int,
                            productvalue numeric(18,2),
                            productweight numeric(18,2),
                            noofitems int,
                            sendername varchar(255),
                            sendercity varchar(100),
                            senderstate varchar(100),
                            receivername varchar(255),
                            receivercity varchar(100),
                            receiverstate varchar(100),
                            estimatedcost numeric(18,2),
                            shipmentno varchar(100),
                            deliverycost numeric(18,2),
                            carriername varchar(255),
                            carriercity varchar(100),
                            carrierstate varchar(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (senderstate) LOCATION '{path}/silver/order';""")
```

```python
def createTable_paymentdetail(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS otracking.silver.paymentdetail
                        (
                            id LONG,
                            serialno VARCHAR(100),
                            paymentdate TIMESTAMP,
                            paidamount numeric(18,2),
                            chequeno varchar(100),
                            refno varchar(100),
                            payingparty varchar(255),
                            receivingparty varchar(255),
                            bankname varchar(100),
                            details varchar(255),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (bankname) LOCATION '{path}/silver/paymentdetail';""")
```

```python
createTable_ordertracking(otracking_path)
createTable_invoice(otracking_path)
createTable_order(otracking_path)
createTable_paymentdetail(otracking_path)
```

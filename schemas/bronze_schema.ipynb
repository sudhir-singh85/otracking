```python
def createTable_ordertracking(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_br.ordertracking
                        (
                            trackingid int,
                            shipmentno VARCHAR(100),
                            orderstatus varchar(100),
                            createdon varchar(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (orderstatus) LOCATION '{path}/ordertracking';""")
```

```python
def createTable_invoice(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_br.invoice
                        (
                            invno VARCHAR(100),
                            invdate VARCHAR(100),
                            ord varchar(100),
                            amount varchar(100),
                            taxrate varchar(100),
                            taxamount varchar(100),
                            paymentdays varchar(100),
                            nameofcustomer varchar(255),
                            city varchar(100),
                            state varchar(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (state) LOCATION '{path}/invoice';""")
```

```python
def createTable_order(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_br.order
                        (
                            ord VARCHAR(100),
                            odate VARCHAR(100),
                            wt varchar(100),
                            ordervalue varchar(100),
                            productcategory varchar(100),
                            productname varchar(100),
                            productsize varchar(100),
                            productvalue varchar(100),
                            productweight varchar(100),
                            noofitems varchar(100),
                            sendername varchar(255),
                            sendercity varchar(100),
                            senderstate varchar(100),
                            receivername varchar(255),
                            receivercity varchar(100),
                            receiverstate varchar(100),
                            estcost varchar(100),
                            shipno varchar(100),
                            dcost varchar(100),
                            carriername varchar(255),
                            carriercity varchar(100),
                            carrierstate varchar(100),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (senderstate) LOCATION '{path}/order';""")
```

```python
def createTable_paymentdetail(path):
    spark.sql(f"""CREATE TABLE IF NOT EXISTS dev.tracking_br.paymentdetail
                        (
                            id VARCHAR(100),
                            srno VARCHAR(100),
                            paymentdate varchar(100),
                            amount varchar(100),
                            cheque varchar(100),
                            refno varchar(100),
                            payingparty varchar(255),
                            receivingparty varchar(255),
                            bankname varchar(100),
                            details varchar(255),
                            ingestion_time TIMESTAMP
                    ) PARTITIONED BY (bankname) LOCATION '{path}/paymentdetail';""")
```

```python
createTable_ordertracking(tracking_br_path)
createTable_invoice(tracking_br_path)
createTable_order(tracking_br_path)
createTable_paymentdetail(tracking_br_path)
```
